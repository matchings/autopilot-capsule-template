# 00 · Data inventory

Discovers every mounted data asset and groups it by modality. Run first to confirm the capsule sees the 5 attached assets.

No Anthropic API key needed.

```python
import sys; sys.path.insert(0,'/root/capsule/code')
from data_loaders import paths
import json
assets=paths.list_assets()
print(f'{len(assets)} assets under {paths.DATA_ROOT}\n')
for a in assets: print(' ',a.name)
```

```python
print(json.dumps(paths.inventory(), indent=2))
```

```python
subjects={}
for a in assets:
    sid=paths.subject_id_from_name(a.name)
    if sid: subjects.setdefault(sid,[]).append(a.name)
for sid,ns in sorted(subjects.items()): print(sid, len(ns))
```
